**1. Створення масивів**

Створіть (кожен пункт оцінюється в 0.5 бала):
- Масив зі списку `[5, 15, 25, 35, 45]`
- Масив з нулів розміром 7
- Масив з одиниць розміром 6
- Масив з числами від 10 до 19 використовуючи `np.arange()`

```python
import numpy as np
list1 = np.array([5, 15, 25, 35, 45])
print(f"Масив зі списку {list1}")
list2 = np.zeros(7)
print(f"Масив з нулів розміром 7: {list2}")
list3 = np.ones(7)
print(f"Масив з одиниць розміром 6: {list3}")
list4 = np.arange(10,20)
print(f"Масив з числами від 10 до 19: {list4}")
```

**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
massyv = np.array([12, 24, 36, 48, 60])
print(f"Numpy-масив зі значеннями {massyv}")
mult = massyv * 2
print(f"Помножте всі елементи на 2 {mult}")
add10 = massyv + 10
print(f"Додайте до всіх елементів 10 {add10}")
squared = massyv ** 2
print(f"Піднесіть всі елементи до квадрату {squared}")
squared_root = np.sqrt(massyv)
print(f"Обчисліть квадратний корінь з усіх елементів {squared_root}")
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
data = np.array([18, 27, 33, 41, 52, 68, 79])
print(f"Numpy-масив зі значеннями {data}")
mean_data = np.mean(data)
print(f"Середнє значення: {mean_data}")
min_data = np.min(data)
print(f"Мінімум: {min_data}")
max_data = np.max(data)
print(f"Максимум: {max_data}")
sum_data = np.sum(data)
print(f"Сума всіх елементів: {sum_data}")
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3 включно
- Елементи більші за 40

```python
arr = np.array([15, 22, 35, 44, 58, 67])
print(f"Numpy-масив зі значеннями {arr}")
first_el = arr[0]
print(f"Перший елемент: {first_el}")
last_el = arr[-1]
print(f"Останній елемент: {last_el}")
el_index = arr[1:4]
print(f"Елементи з індексами 1 до 3 включно: {el_index}")
el_more_40 = arr[arr>40]
print(f"Елементи більші за 40: {el_more_40}")
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
print(f"Форма матриці: {matrix.shape}")
print(f"Перший рядок: {matrix[0,:]}")
print(f"Друга колонка: {matrix[:,1]}")
print(f"Сума всіх елементів: {matrix.sum()}")
print(f"Середнє всіх елементів: {matrix.mean()}")

```
