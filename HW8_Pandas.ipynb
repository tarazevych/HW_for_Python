**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
print("Середній розмір рахунку на всьому наборі даних")
print(df["total_bill"].mean())
print("Максимальні чайові")
print(df["tip"].max())
print("Кількість пропущених значень у всіх колонках")
print(df.isna().sum())
print("Скільки в наборі даних рахунків за обіди і за вечері")
print(df["time"].value_counts())
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
print("Середній рахунок за статтю")
print(df.groupby("sex")["total_bill"].mean())
print("Середні чайові за днем тижня")
print(df.groupby("day")["tip"].mean())
print("Скільки відвідувачів у кожен день")
print(df.groupby("day")["total_bill"].count())
```

**7. Сортування**

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):
- За розміром рахунку (зростання)

Додатково виведіть топ-5 найбільших чайових використовуючи метод `nlargest`.

```python
sorted_bill = df.sort_values(by = "total_bill", ascending = True)
print("За розміром рахунку (зростання)")
print(sorted_bill)
sorted_tip = df.sort_values(by = "tip")
print("За чайовими (спадання)")
print(sorted_tip)
top_5 = df.nlargest(5, "tip")
print("топ-5 найбільших чайових ")
print(top_5)
```
